```python

import pandas as pd

## Load data (kick_1) from a CSV file ##

kick = pd.read_csv('kick_1.csv') # There are mixed data types in columns
```

```python

# Show all columns information #
kick.info()

# View headings of kick_1 #
pd.set_option('display.max_columns', None) # Display all columns. # 
kick.head()
```

```python

kick.shape # Rows, columns #
kick.columns # Names of columns. # 
kick.info() # Data types #
kick.describe() # Overview of numerical columns. #

# Missing values, percentages and rank in kick_1 # 

missing = (                                  # What values are missing. #
    kick.isnull()
        .sum()
        .to_frame(name='Missing Values')
)

missing['Percentage Missing'] = (   # What percentage values are missing. #
    missing['Missing Values'] / len(kick) * 100
).round(2)

missing = missing[                  # Only show volumns with missing variables. #
    missing['Missing Values'] > 0
]

missing = (                                 # Ascensding oreder of values missing. #
    missing.sort_values(
        'Missing Values',
        ascending=False

    )
)
display(missing) # Displqy missinf values, percentages and rank in kink_1. #
```

```python

# Check for duplicate records in kick_1 #

duplicates  = kick.duplicated().sum() 
display(duplicates)

# Check for records that are missing the same values. #

kick[kick.isnull().sum(axis=1) > 10]
```

```python

# Check class distribution & percentage

print(kick['IsBadBuy'].value_counts())
print(kick['IsBadBuy'].value_counts(normalize=True) * 100)
```

```python

# Summary statistics of numerical columns in kick_1 #

kick.describe().round(2) # Round to 2 decimal places. # 

# The number of unique values in each column in kick_1 #

cardinality = (
    kick.nunique()
        .sort_values(ascending=False) # In descending order #
        .reset_index()
)

cardinality.columns = ['Variable', 'Unique Values']
display(cardinality) # Display the number of unique values in each column. #
```
